# Exercise 1: Simulation 2: Python Pin Factory & Punch card

```python
import random

def cut_wire():
    print("Step finished: Wire cut")
    return "Wire cut"

def sharpen_pin():
    print("Step finished: Pin sharpened")
    return "Pin sharpened"

def make_head():
    print("Step finished: Head made")
    return "Head made"

def assemble_pin(components):
    if all(component in components for component in ["Wire cut", "Pin sharpened", "Head made"]):
        print("Step finished: Pin assembled")
        return "Pin assembled"
    else:
        print("Step finished: Incomplete pin")
        return "Incomplete pin"

def create_punch_card(num_rows):
    punch_card = [[1, 1, 1]]  # Ensure at least one row with all 1s
    for _ in range(num_rows - 1):
        punch_card.append(random.choices([0, 1], k=3))
    random.shuffle(punch_card)  # Shuffle to randomize the position of the 111 row
    return punch_card

def execute_punch_card(punch_card):
    pins = []
    for row in punch_card:
        components = []
        if row[0] == 1:
            components.append(cut_wire())
        if row[1] == 1:
            components.append(sharpen_pin())
        if row[2] == 1:
            components.append(make_head())
        pins.append(assemble_pin(components))
    return pins

def main():
    num_rows = 5  # Number of punch card rows (pins to produce)
    punch_card = create_punch_card(num_rows)

    print("Punch Card:")
    for row in punch_card:
        print(row)

    print("\nExecuting Punch Card:")
    pins = execute_punch_card(punch_card)

    print("\nResults:")
    for i, pin in enumerate(pins, 1):
        print(f"Pin {i}: {pin}")

if __name__ == "__main__":
    main()
```

# Exercise 2: Conditional Branching

```python
import random

def upload_document():
    name = input("Enter document name: ")
    doc_type = input("Enter document type (report/proposal/other): ").lower()
    return name, doc_type

def review_document(doc_type):
    if doc_type == "report":
        return review_by_research()
    elif doc_type == "proposal":
        return review_by_finance() and review_by_legal()
    else:
        return review_by_general_manager()

def review_by_research():
    print("Research department reviewing...")
    return random.random() > 0.6  # 40% chance of rejection

def review_by_finance():
    print("Finance department reviewing...")
    return random.random() > 0.5  # 50% chance of rejection

def review_by_legal():
    print("Legal department reviewing...")
    return random.random() > 0.5  # 50% chance of rejection

def review_by_general_manager():
    print("General Manager reviewing...")
    return random.random() > 0.7  # 30% chance of rejection

def process_document():
    name, doc_type = upload_document()
    revision_count = 0
    max_revisions = 2  # Reduced from 3 to 2

    while revision_count < max_revisions:
        print(f"\nProcessing {name} (Type: {doc_type})")

        if review_document(doc_type):
            print(f"{name} approved and added to repository.")
            return
        else:
            revision_count += 1
            if revision_count < max_revisions:
                print(f"{name} needs revision. Attempt {revision_count} of {max_revisions}.")
                input("Press Enter after making revisions...")
            else:
                print(f"{name} rejected after {max_revisions} revision attempts.")
                return

def main():
    while True:
        process_document()
        if input("\nProcess another document? (y/n): ").lower() != 'y':
            break

if __name__ == "__main__":
    main()
```
