**Important Note**: You’ll analyze correlations and regression directly on the bit strings to simulate the "black box" nature of Shannon's approach

**Dataset**  
We will use the "Auto Dataset" (df) from Seaborn, which includes information about various car models. The dataset contains both categorical and quantitative variables. Some relevant columns include:

*   **mpg**: miles per gallon (continuous)
*   **cylinders**: number of cylinders (categorical)
*   **displacement**: engine displacement (continuous)
*   **horsepower**: engine power (continuous)
*   **weight**: weight of the car (continuous)

#### 1. Load and Display the Full Dataset Before Conversion  
   
In this dataset loading step, the dropna() method is used, which removes any rows with missing values (NaN). The original mpg dataset has 398 rows, but 6 rows contain missing values, and these are removed by dropna(), leaving us with 392 rows. Thus, the final dataset after dropping missing values contains 392 rows

```python
import seaborn as sns

# Load the dataset
df = sns.load_dataset('mpg').dropna()

# Display the full dataset
print("Original Dataset:")
print(df)
```

### 2. Convert the Entire Dataset to Bit Strings

```python
import pandas as pd

# Function to convert a value into a bit string representation
def to_bit_string(val):
    if isinstance(val, (int, float)):  # For numerical data
        return format(int(val), '08b')  # Convert to 8-bit binary
    else:  # For categorical data (or strings)
        return format(abs(hash(val)) % (1 << 8), '08b')  # Hash to 8-bit binary

# Apply the conversion function to the entire dataset
bit_string_df = df.applymap(to_bit_string)

# Display the dataset as bit strings
print("\nDataset After Conversion to Bit Strings:")
print(bit_string_df)
```

### 3. Correlation Analysis on Bit Strings   
Here, we convert the bit strings into integers purely for the purpose of performing correlation analysis (since correlations require numerical input). The resulting correlation matrix shows the relationship between the bit strings of different features.  
      
**Explanation**: Correlation is a measure of the strength and direction of a linear relationship between two variables. A correlation coefficient ranges from -1 to 1:
*   -1 indicates a perfect negative linear relationship
*   1 indicates a perfect positive linear relationship
*   0 indicates no linear **relationship**

```python
import numpy as np

# Convert bit strings to integers for correlation analysis
bit_numeric_df = bit_string_df.applymap(lambda x: int(x, 2))

# Compute the correlation matrix
correlation_matrix = bit_numeric_df.corr()

# Display the correlation matrix
print("\nCorrelation Matrix of Bit Strings:")
print(correlation_matrix)
```

### 3. Perform Regression on Bit Strings  
   

This step performs regression analysis where **all the bit strings are used as independent variables**, and the **bit string of mpg is the dependent variable**. We don’t know the meaning of any of the variables, as they are all processed non-symbolically.   
   
**Explanation**  
*  **Coef (Coefficient)**: This is the slope of the relationship between the independent variable (in our case, the bit strings) and the dependent variable (bit string of mpg). It tells us how much the dependent variable changes when the independent variable increases by 1 unit.

*  **Std Err (Standard Error)**: This measures the accuracy of the coefficient’s estimate. A smaller standard error indicates more precise estimates.

*  **t**: The t-statistic is the ratio of the coefficient to its standard error. It is used to test if the coefficient is significantly different from 0. The larger the t-value, the more likely the coefficient is statistically significant.

*  **P>|t| (p-value)**: The p-value tests the null hypothesis that the coefficient is zero (no effect). A small p-value (typically less than 0.05) suggests that the coefficient is statistically significant.

*  **[0.025 0.975]**: This is the 95% confidence interval for the coefficient. It means that we are 95% confident that the true value of the coefficient lies within this range.



```python
import statsmodels.api as sm

# Use all columns except 'mpg' as independent variables, and 'mpg' as the dependent variable
X = bit_numeric_df.drop(columns='mpg')  # Independent variables
y = bit_numeric_df['mpg']  # Dependent variable (bit string of 'mpg')

# Add a constant to X for the regression intercept
X = sm.add_constant(X)

# Perform the regression
model = sm.OLS(y, X).fit()

# Display the regression summary
print("\nRegression Results:")
print(model.summary())
```

### Visualize the Regression Results  
In this step, you will see multiple subplots displaying the regression analysis for different variables (horsepower, weight, displacement), where:

*  **Data Points** (Blue Dots): Represent the bit string values of each feature versus mpg.
*  **Regression Line** (Red Line): Fits the data points, showing the pattern of the relationship between the feature's bit strings and mpg bit strings.
*  **Equation and R-squared Value**: Display the linear regression equation and the R² value, indicating the goodness of fit.
*   **Task**: Plot the data points and regression line. Include the regression equation and R-squared value   
   
**Explanation:**
*   **Data Points**: Represent individual observations
*   **Regression Line**: Shows the best-fit line through the data points
*   **Equation**: MPG = Slope × Horsepower + Intercept

```python
import matplotlib.pyplot as plt

# Select three features for regression (we'll select arbitrarily since we're in Shannon's paradigm)
features = ['horsepower', 'weight', 'displacement']

# Create subplots for visualizing regression
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

for i, feature in enumerate(features):
    # Prepare data
    X_single = bit_numeric_df[feature]
    X_single = sm.add_constant(X_single)

    # Fit regression model
    model_single = sm.OLS(y, X_single).fit()

    # Plot data points and regression line
    axes[i].scatter(bit_numeric_df[feature], bit_numeric_df['mpg'], color='blue', alpha=0.5, label='Data Points')
    axes[i].plot(bit_numeric_df[feature], model_single.predict(X_single), color='red', label='Regression Line')

    # Display equation and R-squared value on each plot
    equation = f"y = {model_single.params[1]:.4f}x + {model_single.params[0]:.4f}"
    r_squared = model_single.rsquared
    axes[i].text(0.05, 0.95, f"Eq: {equation}\nR²: {r_squared:.4f}", transform=axes[i].transAxes, fontsize=12,
                 verticalalignment='top', bbox=dict(facecolor='white', alpha=0.8))

    # Set labels and title
    axes[i].set_title(f'Regression: {feature} vs MPG')
    axes[i].set_xlabel(f'{feature} (Bit String)')
    axes[i].set_ylabel('MPG (Bit String)')
    axes[i].legend()

# Display all subplots
plt.tight_layout()
plt.show()
```
